```python
import pandas as pd
import numpy as np
```

# Find missing value

```python
x = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(x)
```

```python
df
```

```python
df.isna()
```

```python
df.isna().sum() # number of null value in cols
```

```python
df.isna().any()
```

# Drop Missing row

```python
df.dropna() # if find 1 null in the row, then drop it
```

```python
df.dropna(thresh=3) # at least 3 value, or drop
```

# Fill Missing Value

```python
df.fillna(0)
```

```python
vals = {'A': 100, 'B': 200, 'C': 300, 'D': 400}
df.fillna(value=vals)
```

```python
df.fillna(df.mean())
```
