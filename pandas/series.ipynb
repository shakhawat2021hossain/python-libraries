# Series
- Series is one-dimensional labeled array
- can be created from array, list, dictionary 

```python
x = pd.Series([1, 2, 3, 4, 5])
print("series:", x)
print(f"index: {x.index}, \nvalues: {x.values}")
print(f"max: {x.max()}, min: {x.min()}, sum: {x.sum()}")
```

```python
x.describe()
```

```python
# with pre defined index
x = pd.Series([1, 2, 3], index=['a', 'b', 'c'])
```
