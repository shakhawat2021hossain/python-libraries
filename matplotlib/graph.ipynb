```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(0, 100)

y1 = x**2
y2 = x**3

# plt.plot(x, y1, color="red")
plt.plot(x, y2)

plt.grid()
plt.show()

```

```python

x = np.linspace(0, 2*np.pi, 100)
# print(x)
y1 = np.sin(x)
y2 = np.cos(x)

plt.plot(x, y1)
plt.plot(x, y2)
plt.xlabel("value of x")
plt.ylabel("amplitude")
plt.grid()
plt.show()

```
