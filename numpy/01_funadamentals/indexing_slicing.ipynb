# Indexing: Accessing a particular element from an array
# Slicing: Taking a portion from an array

```python
# indexing in normal python list

x = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
print(x[0])
print(x[0][2]) 
```

```python
# indexing in numpy array

import numpy as np
arr = np.array(x)

print(arr[0, 2]) # arr[row, col]
```
