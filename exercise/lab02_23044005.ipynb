### Exercise 1: Aircraft Maintenance Telemetry Analysis with NumPy Arrays ✈️

You are part of an **aircraft maintenance engineering team** responsible for monitoring aircraft systems and analyzing sensor data during a test flight. The maintenance system requires efficient handling of numerical telemetry data to identify potential performance issues.

#### Objectives:

* Generate a NumPy array representing **flight time** in seconds from `t=0` to `t=100` seconds with a step size of `1` second.

* Generate a NumPy array representing **engine temperature** in °C. Assume the temperature starts at `80°C` and increases linearly at a rate of `0.5°C` per second during the test.

* Generate a NumPy array representing **engine vibration level** in mm/s. Assume the vibration level starts at `2 mm/s` and increases by `0.05 mm/s` per second.

* Generate an **acceleration array** representing a constant aircraft acceleration of `0.2 m/s²` throughout the test.

* Print the **shape** and **data type** of each generated NumPy array.


```python
import numpy as np

flight_time = np.arange(0, 101, 1)
engine_temperature = 80 + 0.5 * flight_time
engine_vibration_level = 2 + 0.05 * flight_time
acceleration = np.full(flight_time.shape, 0.2)

print(f"flight_time array shape: {flight_time.shape}, data type: {flight_time.dtype}")
print(f"engine_temperature array shape: {engine_temperature.shape}, data type: {engine_temperature.dtype}")
print(f"engine_vibration_level array shape: {engine_vibration_level.shape}, data type: {engine_vibration_level.dtype}")
print(f"acceleration array shape: {acceleration.shape}, data type: {acceleration.dtype}")
```

### Exercise 2: Plotting Mathematical Functions 🧮 
Let's now apply our further understanding of `math`, `matplotlib`, and `numpy` libraries to plot some nifty mathematical functions.

#### Objectives:
- Initialize a line plot using `matplotlib`
- Create a plot of a `tan` function with names for the plot title, x-axis, and y-axis of your choice
- Create a plot with a `tan` and `cos` function with names for the plot title, x-axis, and y-axis of your choice
- For both plots, ensure the x-axis has equal spacing from 0 to 2pi using 50 samples
- Make sure both plots have a grid enabled

```python
import matplotlib.pyplot as plt


x = np.linspace(0, 2 * np.pi, 50)
y1 = np.tan(x)
y2 = np.cos(x)

# graph 1
plt.plot(x, y1, label='tan(x)')
plt.title('tan(x) graph')
plt.xlabel('value of x (radians)')
plt.ylabel('Amplitude')
plt.legend()
plt.grid()
plt.show()

# graph 2

plt.plot(x, y2, label='cos(x)')
plt.plot(x, y1, label='tan(x)')
plt.title('tan(x) and cos(x) graph')
plt.xlabel('value of x (radians)')
plt.ylabel('Amplitude')
plt.legend()
plt.grid()
plt.show()

```
