### Exercise 1: Aircraft Maintenance Telemetry Analysis with NumPy Arrays ✈️

You are part of an **aircraft maintenance engineering team** responsible for monitoring aircraft systems and analyzing sensor data during a test flight. The maintenance system requires efficient handling of numerical telemetry data to identify potential performance issues.

#### Objectives:

* Generate a NumPy array representing **flight time** in seconds from `t=0` to `t=100` seconds with a step size of `1` second.

* Generate a NumPy array representing **engine temperature** in °C. Assume the temperature starts at `80°C` and increases linearly at a rate of `0.5°C` per second during the test.

* Generate a NumPy array representing **engine vibration level** in mm/s. Assume the vibration level starts at `2 mm/s` and increases by `0.05 mm/s` per second.

* Generate an **acceleration array** representing a constant aircraft acceleration of `0.2 m/s²` throughout the test.

* Print the **shape** and **data type** of each generated NumPy array.
